```python
import json
import os
import re

NUM_WINDOWS = 4
WINDOW_OFFSET = 15 # The window offset in minutes in which the geometry remains fairly consistent
SAMPLES_DIR = os.path.join(os.getcwd(), "..", "..", "gnss-data-collection",
                           "window-experiment", "data", "samples")
GEOM_DIR = os.path.join(os.getcwd(), "..", "geometry-consistent-data")
```

```python
exp = r"^c3.2_day\d{1}_manifest.json$" # This captures c3.2 data across different days

# Helps sort by number for data collected across days, keeping geometry consistent
def sort_by_number(filename):
    return [text.lower() for text in re.split(r"(\d+)", filename)]

c32_manifest_files = sorted([
    filename for filename in os.listdir(SAMPLES_DIR)
    if re.search(exp, filename)
], key=sort_by_number)

data = []
for manifest in c32_manifest_files:
    with open(os.path.join(SAMPLES_DIR, manifest), "r") as m:
        data.append(json.load(m))
```

```python
# Each window has consistent geometry
geom_consistent_data = []

for i in range(0, NUM_WINDOWS):
    window_data = []

    for j in range(0, len(data)):
        entries = data[j]["entries"]

        for k in range(0, len(entries)):
            if entries[k]["window_index"] == i:
                window_data.append((entries[k]["rtcm"], entries[k]["meta"]))

    geom_consistent_data.append(window_data)
```

```python
with open(os.path.join(GEOM_DIR, "gc_c3.2_data1.json"), "w") as out:
    json.dump(geom_consistent_data, out)
```
